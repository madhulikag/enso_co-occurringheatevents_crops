```python
import pandas as pd
import matplotlib.pyplot as plt

# Define the data
data = {
    "Season": ['JJA(0)', 'SON(0)', 'D(0)JF(+1)', 'MAM(+1)', 'JJA(+1)', 'SON(+1)'],
    "ElNino_Minus_Neutral_CroplandSignificantPosArea_ha": [31798266, 36506888, 36038696, 34198872, 58223212, 27248548],
    "LaNina_Minus_Neutral_CroplandSignificantPosArea_ha": [15324655, 8529757, 12240650, 7163299.50, 7983285.50, 3749788],
    "ElNino_Minus_Neutral_CroplandSignificantNegArea_ha": [-853495.69, -1559178.25, -25309476, -4245266, -914688.31, -3761713],
    "LaNina_Minus_Neutral_CroplandSignificantNegArea_ha": [-1906949.50, -2505230, -12745605, -6866112.50, -1279526.88, -3266151.50]
}

# Create the DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

```python
# Plot
fig, ax = plt.subplots(figsize=(10, 6))
bar_width = 0.35
x = range(len(df))

ax.bar(x, df['ElNino_Minus_Neutral_CroplandSignificantPosArea_ha'], width=bar_width, label='El Niño - Neutral', color='#E24A33')
ax.bar([p + bar_width for p in x], df['LaNina_Minus_Neutral_CroplandSignificantPosArea_ha'], width=bar_width, label='La Niña - Neutral', color='#348ABD')

# Labels and formatting
ax.set_xticks([p + bar_width / 2 for p in x])
ax.set_xticklabels(df['Season'], fontsize=18)
ax.set_ylabel('Cropland Area (ha)', fontsize=18)
ax.set_title('Cropland area (ha) exposed to higher likelihood of Hot-Wet events', fontsize=20)
ax.tick_params(axis='y', labelsize=18)  # Increase y-axis tick label font size
ax.legend(fontsize=18)  # Increase legend font size

plt.tight_layout()
plt.show()
```

```python
# Plotting
fig, ax = plt.subplots(figsize=(12, 7))
bar_width = 0.35
x = range(len(df))

# Plot positive and negative bars for El Niño
elnino_pos = ax.bar(x, df['ElNino_Minus_Neutral_CroplandSignificantPosArea_ha'], width=bar_width, label='El Niño - Neutral', color='#E24A33')
elnino_neg = ax.bar(x, df['ElNino_Minus_Neutral_CroplandSignificantNegArea_ha'], width=bar_width, color='#E24A33')

for i in [0, 3, 4]:
    elnino_pos[i].set_hatch('//')  # hatch pattern
    

for i in [2]:
    elnino_neg[i].set_hatch('//')  # hatch pattern
    
# Plot positive and negative bars for La Niña
lanina_pos = ax.bar([p + bar_width for p in x], df['LaNina_Minus_Neutral_CroplandSignificantPosArea_ha'], width=bar_width, label='La Niña - Neutral', color='#348ABD')
lanina_neg = ax.bar([p + bar_width for p in x], df['LaNina_Minus_Neutral_CroplandSignificantNegArea_ha'], width=bar_width, color='#348ABD')

# for i in [1]:
#     lanina_pos[i].set_hatch('//')  # hatch pattern
    
# Formatting
ax.set_xticks([p + bar_width / 2 for p in x])
ax.set_xticklabels(df['Season'], fontsize=18)
ax.set_ylabel('Cropland Area (ha)', fontsize=18)
ax.set_title('Cropland Area (ha) with Significant Hot-Wet Likelihood', fontsize=20)
ax.tick_params(axis='y', labelsize=18)
ax.yaxis.get_offset_text().set_fontsize(18) 
ax.axhline(0, color='black', linewidth=0.8)
# ax.legend(fontsize=18, loc='lower right')

plt.tight_layout()
plt.show()
```

```python
# Normalize values as percentage of 1710239104
denominator = 1710239104
df["ElNino_Percent_pos"] = df["ElNino_Minus_Neutral_CroplandSignificantPosArea_ha"]*100 / denominator 
df["LaNina_Percent_pos"] = df["LaNina_Minus_Neutral_CroplandSignificantPosArea_ha"]*100 / denominator 
df["ElNino_Percent_neg"] = df["ElNino_Minus_Neutral_CroplandSignificantNegArea_ha"]*100 / denominator 
df["LaNina_Percent_neg"] = df["LaNina_Minus_Neutral_CroplandSignificantNegArea_ha"]*100 / denominator 
```
