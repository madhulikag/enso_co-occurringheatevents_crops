```python
## This code reconstructs the 1870-2025 Niño 3.4 anomaly time series to the 1981–2010 monthly climatology

import pandas as pd
import numpy as np

# data for nino3.4 anomaly 1870-2025 based on 1981-2010 climatology is from https://psl.noaa.gov/data/timeseries/month/data/nino34.long.anom.data

file_path = '/home/madhulika.gurazada/paper2/Data/Nino3.4anom.xlsx'

nino34anom = pd.read_excel(file_path)

# climatology data is from https://ftp.cpc.ncep.noaa.gov/wd52dg/data/indices/ersst3b.nino.mth.81-10.ascii

file_path = '/home/madhulika.gurazada/paper2/Data/Nino3.4clim.xlsx'

nino34clim = pd.read_excel(file_path)

nino34clim = nino34clim.rename(columns={
    'YR': 'Year',
    'MON': 'Month',
    'NINO3.4': 'nino3.4'
})

nino34clim = nino34clim.drop(columns=['ANOM'])

nino34clim = nino34clim[(nino34clim.Year >= 1981) & (nino34clim.Year <= 2010)]

nino34clim = (
    nino34clim
    .groupby('Month')['nino3.4']
    .mean()
    .rename('climatology')        
)

months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']

nino34clim = pd.Series(data=nino34clim.values, index=months)

nino34anom[months] = nino34anom[months].replace(-99.99, np.nan)

nino34sstraw = nino34anom.copy()

nino34sstraw[months] = nino34anom[months].add(nino34clim, axis=1)

nino34sstraw.to_excel(
    '/home/madhulika.gurazada/paper2/Data/Nino3.4_OG_SST1.xlsx',
    index=False,
    engine='openpyxl'
)
```
